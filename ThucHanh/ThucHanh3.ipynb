B5: giải phương trình a*x**2 + b*x + c = 0

```python
def ptb2(a, b, c):
    print(f"Phương trình: {a}x^2 + {b}x + {c} = 0")
    if a == 0:
        if b == 0:
            if c == 0:
                return "vô số nghiệm"
            else: 
                return "vô nghiệm"
        else:
            x = -c/b
            return f"Nghiệm của phương trình ax + b = 0 là: x = {x:.2f}"
    d = b**2 - 4*a*c
    if d < 0:
        return "vô nghiệm"
    elif d == 0:
        x = -b/(2*a)
        return f"Phương trình có nghiệm kép: x = {x:.2f}"
    else:
        x1 = (-b + d**0.5)/(2*a)
        x2 = (-b - d**0.5)/(2*a)
        return f"Phương trình có hai nghiệm phân biệt: x1 = {x1:.2f}, x2 = {x2:.2f}"
a, b, c = map(float, input().split())
print(ptb2(a, b, c))
```
