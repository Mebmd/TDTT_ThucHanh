16: ktra số Amstronhg? Hàm trả về kiểu bool. vd 153 = 1^3 + 5^3 + 3^3

```python
def Armstrong(x: int):
    tong = 0
    n = len(str(x))
    thu = x
    while thu > 0:
        tong += (thu%10)**n
        thu //= 10
    if tong == x:
        return True
    else:
        return False
x = int(input())
print(Armstrong(x))
```

17: ktra số chính phương

```python
def ktra(a):
    if (a**0.5) %1 == 0:
        return True
    else:
        return False
a = int(input())
print(ktra(a))
```

18: kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên

```python
def sodx(x):
    nguoc = str(x)[::-1]
    if int(nguoc) == x:
        return True
    else:
        return False
x = int(input())
print(sodx(x))
```
