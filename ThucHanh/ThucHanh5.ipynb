25: Tìm số lớn nhất và nhỏ nhất trong dãy số (dừng khi nhập -1) 

```python
n = int(input())
if n == -1:
    print("dung lai")
else:
    max_val = n
    min_val = n
    while True:
        n = int(input())
        if n == -1:
            break
        if n > max_val:
            max_val = n
        if n < min_val:
            min_val = n
            
    print(f"So lon nhat la: {max_val}")
    print(f"So nho nhat la:{min_val}")
```

26: Số Fibonacci lớn nhất không vượt quá A

```python
A = int(input())
f1 = 1
f2 = 1
if A < 1:
    print("nhap lai")
else:
    while f1 + f2 <= A:
        f_next = f1 + f2
        f1 = f2
        f2 = f_next
    print(f"{f2}")
```

```python
def tim(lst, k: int):
    n = len(lst)
    for i in range(n):
        if lst[i] == k:
            return i
    return -1
lst = list(map(int, input().split()))
k = int(input())
print(tim(lst, k))
```

8: khoảng cách Hamming giữa 2 số a và b

```python
def hamming_distance(a: int, b: int):
    kcach = 0
    while a>0 or b>0:
        bita = a%2
        bitb = b%2
        if bita != bitb:
            kcach += 1
        a //= 2
        b //= 2
    return kcach
a, b = map(int, input().split())
print(hamming_distance(a, b))
```

10: viết hàm trả về true nếu a, b đẳng cấu, ngược lại trả về false

```python
def dang_cau(a,b):
    if len(a) != len(b):
        return False
    ab = {}
    ba = {}
    for m, n in zip(a,b): 
        if m in ab:
           if ab[m] != n:
                return False
        else:
            ab[m] = n
        if n in ba:
            if ba[n] != m:
                return False
        else:
            ba[n] = m
    return True
a,b = map(str,input().split())
print(dang_cau(a,b))
```
