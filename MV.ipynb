```python
nums = list(map(int, input().split()))
tongam = 0
cosoam = False

tongduong = 0
demduong = 0

lonnhat = nums[0]
nhonhat = nums[0]

for x in nums:
    if x < 0:
        tongam += x
        cosoam = True
    if x > 0:
        tongduong += x
        demduong += 1
    if x > lonnhat:
        lonnhat = x
    if x < nhonhat:
        nhonhat = x
tongam = tongam if cosoam else "NaN"
if demduong > 0:
    tbduong = tongduong/demduong
else:
    tbduong = "NaN"
print(f"{tongam}\t{tbduong:.2f}\t{lonnhat}\t{nhonhat}")
    
```

```python
a, b = map (int, input().split())
tonga = 1
tongb = 1
for i in range(2, a):
    if a % i == 0:
        tonga += i
for i in range(2, b):
    if b % i == 0:
        tongb += i
if tonga == b and tongb == a:
    print("True")
else:  
    print("False")
    
```

```python
n = int(input())
if n <= 1:
    print("Invalid")
else:
    d = 2
    c = n
    while c > 1:
        if c % d == 0:
            lanchia = 0
            while c % d == 0:
                lanchia += 1
                c //= d
            print(f"{d}^{lanchia}",sep="" , end=" ")
            if c > 1:
                print("*", end="")
        d += 1
```
